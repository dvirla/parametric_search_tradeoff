```python
from datasets import load_dataset
```

```python
ds = load_dataset("katielink/expertqa", split="train")
```

```python
ds
```

```python
set(ds['metadata.field'])
```

```python
question_types = set()
for q_type in ds['metadata.question_type']:
    for t in q_type.split('|'):
        question_types.add(t.strip())
question_types
```

```python
filtered_ds = ds.filter(lambda example: "Directed question that has a single unambiguous answer" in example["metadata.question_type"])
```

```python
len(set(filtered_ds['metadata.field']))
```

```python
filtered_df = filtered_ds.to_pandas()
filtered_df['metadata.field'].value_counts()
```

```python
df_frac_sample = filtered_df.groupby('metadata.field').apply(lambda x: x.sample(frac=0.1, random_state=42))
```

```python
df_frac_sample.reset_index().drop(columns=['level_1'], inplace=True)
```

```python
df_frac_sample.reset_index().drop(columns=['level_1']).to_csv('../data/expertqa_sample.csv', index=False)
```
